# Datum — Yahoo Finance & Binance Example
Pulls daily OHLCV for **AAPL** (Yahoo) and **BTCUSDT** (Binance) for March 2026.

```python
import sys
import pathlib

# Add src/ to path so datum can be imported without installing the package
sys.path.insert(0, str(pathlib.Path().resolve().parent / "src"))
```

```python
from datum import Datum
```

```python
import yfinance as yf

raw_yf = yf.download(
    tickers=["AAPL"],
    start="2026-03-01",
    end="2026-03-31",
    interval="1d",
    auto_adjust=False,
    actions=False,
    group_by="ticker",
    threads=False,
    progress=False,
)

# Flatten to match what Datum produces for easy comparison
raw_yf.columns = raw_yf.columns.droplevel(0)          # drop ticker level
raw_yf = raw_yf[["Open", "High", "Low", "Close", "Volume"]].rename(
    columns=str.lower
)
raw_yf
```

## Yahoo Finance — AAPL

```python
d_equity = Datum(
    tickers=["AAPL"],
    start="2026-03-01",
    end="2026-03-31",
    freq="1d",
    provider="yahoo",
    dropna_policy="forward",
)

prices_equity = d_equity.get_prices()
prices_equity
```

```python
d_equity.get_metadata()
```

## Binance — BTCUSDT

```python
d_crypto = Datum(
    tickers=["BTCUSDT"],
    start="2026-03-01",
    end="2026-03-31",
    freq="1d",
    provider="binance",
    dropna_policy="strict",
)

prices_crypto = d_crypto.get_prices()
prices_crypto
```

```python
d_crypto.get_metadata()
```

## Schema check — both DataFrames have identical field structure

```python
equity_fields = prices_equity.columns.get_level_values(1).unique().tolist()
crypto_fields = prices_crypto.columns.get_level_values(1).unique().tolist()

assert equity_fields == crypto_fields, f"Schema mismatch: {equity_fields} vs {crypto_fields}"
print("Fields match:", equity_fields)
```

## Preprocessor — quality reports & cleaning

`Preprocessor` takes a Datum prices frame and produces:
- **coverage** (start/end/gaps), **missing data** counts, **outlier** flags (z-score on log returns), and **structural quality** checks
- a cleaned OHLCV frame where extreme returns are winsorized to ±threshold·σ

```python
from datum import Preprocessor

pre = Preprocessor(prices_equity, zscore_threshold=4.0)
pre.coverage_report()
```

```python
pre.missing_report()
```

```python
pre.outlier_report()
```

```python
pre.quality_report()
```

```python
cleaned = pre.clean()
cleaned
```
